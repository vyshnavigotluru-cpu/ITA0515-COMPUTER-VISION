```python
!wget -q https://github.com/opencv/opencv_zoo/raw/refs/heads/main/models/face_detection_yunet/face_detection_yunet_2023mar.onnx

import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

uploaded = files.upload()
name = list(uploaded.keys())[0]

img = cv2.imread(name)
h, w = img.shape[:2]

model = cv2.FaceDetectorYN_create(
    "face_detection_yunet_2023mar.onnx",
    "",
    (w, h),
    0.6,
    0.3,
    5000
)

_, faces = model.detect(img)

count = 0

if faces is not None:
    for face in faces:
        x, y, fw, fh = map(int, face[:4])
        cv2.rectangle(img, (x, y), (x+fw, y+fh), (0, 255, 0), 2)
        count += 1

print("Number of Faces:", count)
cv2_imshow(img)
```
